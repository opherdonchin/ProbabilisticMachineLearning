# Sensor fusion schematic

Course-owned schematic inspired by Murphy, *Probabilistic Machine Learning: Advanced Topics*, Fig. 2.10. It illustrates how the posterior mean changes as relative sensor uncertainty changes.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

fig, axes = plt.subplots(1, 3, figsize=(11.5, 3.3), constrained_layout=True)

configs = [
    dict(title='Equal uncertainty', x=(0.25,0.72), y=(0.75,0.28), ex=(0.24,0.24), ey=(0.24,0.24), post=(0.50,0.50), caption='posterior mean lies between them'),
    dict(title='$y$ more precise', x=(0.25,0.72), y=(0.75,0.28), ex=(0.38,0.38), ey=(0.16,0.16), post=(0.66,0.36), caption='estimate shifts toward $y$'),
    dict(title='Precision depends on direction', x=(0.25,0.72), y=(0.75,0.28), ex=(0.40,0.12), ey=(0.12,0.40), post=(0.68,0.66), caption='each direction follows the more precise sensor'),
]

colors = plt.rcParams['axes.prop_cycle'].by_key()['color']
cx, cy = colors[0], colors[1]

for ax, c in zip(axes, configs):
    ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.set_aspect('equal')
    ax.set_xticks([]); ax.set_yticks([])
    for s in ax.spines.values():
        s.set_visible(False)
    ax.set_title(c['title'], fontsize=11, pad=6)

    ax.add_patch(Ellipse(c['x'], *c['ex'], fill=False, linewidth=1.6, edgecolor=cx))
    ax.add_patch(Ellipse(c['y'], *c['ey'], fill=False, linewidth=1.6, edgecolor=cy))
    ax.scatter(*c['x'], marker='x', s=75, linewidths=2.2, color=cx, label='$x$ measurement')
    ax.scatter(*c['y'], marker='x', s=75, linewidths=2.2, color=cy, label='$y$ measurement')
    ax.scatter(*c['post'], marker='D', s=40, color='black', label='posterior mean')
    ax.text(0.50, 0.06, c['caption'], ha='center', va='bottom', fontsize=8.3)

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc='lower center', ncol=3, frameon=False, bbox_to_anchor=(0.5, -0.03), fontsize=9)
fig.savefig('generated/sensor_fusion_schematic.svg', bbox_inches='tight')
fig.savefig('generated/sensor_fusion_schematic.png', dpi=180, bbox_inches='tight')
```
